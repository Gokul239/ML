Developing the perceptron model from scratch

```python
import numpy as np

class Perceptron:

    '''  
    Building the perceptron classifer

    parameters
    -----------

    learning_rate: float
    between(0.0-1.0)

    epochs: int
    number of iteration paases over the training dataset

    random_state: int
    Random number generator seed for random weight initialization


    '''

    def __init__(self, learning_rate=0.01, epochs = 20, random_state=1):

        self.learning_rate = learning_rate
        self.epochs = epochs
        self.random_state = random_state

    def fit(self, x, y):

        self.errors_ = []
        rgen = np.random.RandomState(1)
        self.weights_ = rgen.normal(loc=0, scale=0.01, size=x.shape[1]+1)
        
        for _ in range(self.epochs):
            errors = 0

            for xi, yi in zip(x, y):
                update = self.learning_rate*(yi- self.prediction(xi))

                self.weights_[1: ] += update*xi
                self.weights_[0] += update

                errors+=int(update!=0)
            self.errors_.append(errors)
        return self    

    def prediction(self, x):

        predicted = self.weights_[0]+(np.dot(x, self.weights_[1: ]))
        return self.activation_(predicted)

    def activation_(self, x):
        return np.where(x >= 0, 1, -1)

```

```python
import warnings
warnings.filterwarnings("ignore")
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn import datasets
```

```python
iris = datasets.load_iris()

df = pd.DataFrame(data=iris.data, columns=iris.feature_names)
df['target'] = iris.target
```

```python
df.target.unique()
```

```python
df = df[df.target!=2]
df.sample(frac=1)
```

```python
from sklearn.preprocessing import StandardScaler

sc = StandardScaler()

x = sc.fit_transform(df[iris.feature_names])
```

```python
df.target = df.target.apply(lambda x: -1 if x == 0 else 1)
```

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(x, df['target'], test_size=0.20, random_state=100, stratify=df['target'])
```

```python
perc = Perceptron()
```

```python
perc.fit(x_train, y_train)
```

```python
y_pred = perc.prediction(x_test)
```

```python
from sklearn.metrics import accuracy_score

accuracy_score(y_test, y_pred)
```
